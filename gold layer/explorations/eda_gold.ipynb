```python
%sql
SELECT
    COUNT(DISTINCT customer_id) AS unique_cust_id,
    COUNT(DISTINCT order_id) AS unique_order_id,
    COUNT(DISTINCT product_id) AS unique_product_id,
    COUNT(DISTINCT order_date_id) AS unique_order_date_id
FROM supply_chain_demo.gold.fct_orderlines;
```

```python
%sql
USE CATALOG supply_chain_demo;
USE SCHEMA gold;

SELECT 
    o.order_id,
    o.order_status,
    ol.total_amount,
    p.product_name,
    p.category_name,
    d.datetime,
    d.year,
    d.weekday,
    c.country,
    c.state,
    c.segment
FROM  fct_orderlines ol 
LEFT JOIN dim_customer c ON ol.customer_id = c.customer_id
LEFT JOIN dim_date d ON ol.order_date_id = d.datetime_id
LEFT JOIN dim_order o ON ol.order_id = o.order_id
LEFT JOIN dim_product p ON ol.product_id = p.product_id
WHERE c.country = 'Puerto Rico'
```
